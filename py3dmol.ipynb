```python
import py3Dmol
ibux = py3Dmol.view(query='cid:155903259')
ibux.setStyle({'stick': {'color': 'spectrum'}})
ibux.show()
```

In this cell, we're telling `py3Dmol` to grab the structure of Ibuprofen for us! We're using its special PubChem ID (`155903259`) to find it. Then, we tell it to draw the molecule as 'sticks' and color it in a lovely 'spectrum' of colors based on the different types of atoms.

```python
import py3Dmol

ibux = py3Dmol.view(query='cid:155903259')
ibux.setStyle({'stick': {'color': 'spectrum'}})
ibux.addSurface('VDw', opacity=0.5, colorScheme=('gradient','rwd'))
ibux.show()
```

Building on the last step, we're again showing Ibuprofen as 'sticks' with pretty 'spectrum' colors. But this time, we're also adding a transparent 'surface' around it, like a bubbly force field, to show its overall shape and how much space it takes up. The colors on this surface fade from red to white to blue.

```python
Mpro = py3Dmol.view(query='pdb:7RFS')
Mpro.setStyle({'cartoon': {'color': 'spectrum'}})
Mpro.show()
```

Now we're moving on to a protein! This code fetches the structure of protein namely Main Protease (Mpro) of SARS-CoV-2 using its PDB ID (`7RFS`). We're telling it to draw the protein in a 'cartoon' style, which is super helpful for seeing its main structural features like spirals and ribbons, again with those vibrant 'spectrum' colors.

```python
Mpro = py3Dmol.view(query='pdb:7RFS')
Mpro.setStyle({'cartoon': {'color': 'spectrum'}})
Mpro.addSurface('VDw', opacity=0.1, colorScheme=('gradient','rwd'))
Mpro.show()
```

Finally, we're looking at that Celecoxib protein again in its 'cartoon' glory, but we're adding a very subtle, transparent surface this time (only 10% opaque). This lets us still clearly see the cool cartoon structure underneath while getting a hint of the protein's overall boundary, using the same red-white-blue gradient colors.
