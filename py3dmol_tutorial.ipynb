## Visualization of Protein and Ligand Structures using py3Dmol


This notebook visualizes the ligand Nirmatrelvir (the active antiviral component of Paxlovid) and its target, the SARS-CoV-2 main protease (M^pro), fetched directly from PubChem and the RCSB PDB.

This installs `py3Dmol`, the library used throughout this notebook to render 3D molecular structures inline.


### 1. Ligand visualization: Nirmatrelvir

Nirmatrelvir is fetched directly from PubChem using its Compound ID (CID 155903259) and rendered as a stick model.


```python
import py3Dmol

nmt = py3Dmol.view(query='cid:155903259')  # Nirmatrelvir, PubChem CID 155903259
nmt.setStyle({'stick': {'color': 'spectrum'}})
nmt.show()
```

The structure is retrieved using its PubChem CID (`155903259`, Nirmatrelvir) and rendered in stick representation, colored by spectrum to distinguish individual atoms across the molecule.


```python
import py3Dmol

nmt = py3Dmol.view(query='cid:155903259')  # Nirmatrelvir, PubChem CID 155903259
nmt.setStyle({'stick': {'color': 'spectrum'}})
nmt.addSurface('VDW', opacity=0.5, colorScheme=('gradient', 'rwd'))
nmt.show()
```

Building on the previous view, Nirmatrelvir is again shown in stick representation, with a semi-transparent Van der Waals surface (opacity 0.5) added on top. The surface uses a red-white-blue gradient and approximates the molecule's overall shape and volume, in addition to its atomic connectivity.


```python
Mpro = py3Dmol.view(query='pdb:7RFS')  # SARS-CoV-2 main protease (M^pro)
Mpro.setStyle({'cartoon': {'color': 'spectrum'}})
Mpro.show()
```

### 2. Protein visualization: SARS-CoV-2 main protease (M^pro)

The protein structure is fetched directly using its PDB ID (`7RFS`) and rendered in cartoon representation, colored by spectrum. Cartoon mode highlights secondary structure elements (helices, sheets, loops) rather than individual atoms.


```python
Mpro = py3Dmol.view(query='pdb:7RFS')  # SARS-CoV-2 main protease (M^pro)
Mpro.setStyle({'cartoon': {'color': 'spectrum'}})
Mpro.addSurface('VDW', opacity=0.1, colorScheme=('gradient', 'rwd'))
Mpro.show()
```

A near-transparent Van der Waals surface (opacity 0.1) is added on top of the cartoon structure. A low opacity is used deliberately here: a dense surface would obscure the secondary structure detail that the cartoon representation is meant to show, so this view trades surface clarity for visibility of the underlying fold.
